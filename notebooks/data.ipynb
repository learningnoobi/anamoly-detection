```python
sampled = []
for t, group in df.groupby('type'):
    sampled.append(group.sample(n=min(len(group), 1000), random_state=42))
filtered = pd.concat(sampled, ignore_index=True)
print(filtered['type'].value_counts())  # verify type column is there
```
