```python
import os
from turtle import mode

def combine_or_extract_normal_csv(csv_folder,  mode="normal"):
    """
    Extracts rows with "type" == "normal" from all CSV files in the given folder
    or combines all CSV files without filtering, depending on the mode,
    and saves them to a single output CSV file.

    """
    final_data = []
    total_rows = 0
    all_normal_rows = 0
    output_file = normal_data_csv if mode == "normal" else all_network_data_csv
    print(f"Mode is set to '{mode}'. Output will be saved to {output_file}")
    if os.path.exists(output_file):
        print(f"Output file {output_file} already exists. Please remove it before running the script.")
        return
    
    # Iterate through all files in the folder
    for filename in os.listdir(csv_folder):
        file_path = os.path.join(csv_folder, filename)
        
        # Check if the file is a CSV
        if not filename.endswith('.csv'):
            continue
        try:
            df = pd.read_csv(file_path)
            rows = df.shape[0]
            total_rows += rows
            if mode =="normal":
                # Filter rows where "type" is "normal"
                normal_df = df[df['type'] == 'normal']
                normal_rows = normal_df.shape[0]
                all_normal_rows += normal_rows
                
                # Append to the list of normal data
                final_data.append(normal_df)
                print(f"Processed {filename} - {len(normal_df)} normal rows")
            else:
                # combine all data
                final_data.append(df)
                print(f"Processed {filename} - {len(df)} rows")
        except Exception as e:
            print(f"Error processing {filename}: {e}")

    print(f"\nTotal rows processed: {total_rows}")
    print(f"Total normal rows found: {all_normal_rows}")
    
    # Combine all normal data into one DataFrame
    if final_data:
        combined_df = pd.concat(final_data, ignore_index=True)
        print("trying to save final on ",output_file)
        combined_df.to_csv(output_file, index=False)
        print(f"\nSaved {len(combined_df)} normal rows to {output_file}")
    else:
        print("No CSV files found or no  rows to save.")

# Example usage:
# combine_or_extract_normal_csv(csv_folder)
combine_or_extract_normal_csv(csv_folder, mode="all")
```

```python
def combine_or_extract_normal_csv2(csv_folder, mode="normal", max_rows=1_000_000):
    final_data = []
    total_rows = 0
    output_file = normal_data_csv if mode == "normal" else all_network_data_csv

    print(f"Mode: '{mode}' | Output: {output_file}")

    # for filename in os.listdir(csv_folder):
    #     if not filename.endswith('.csv'):
    #         continue
    #     df = pd.read_csv(csv_folder + '/' + filename, nrows=2)
    #     print(f"{filename}: {sorted(df.columns.tolist())}")

    # return

    if os.path.exists(output_file):
        print(f"Output file already exists. Remove it before running.")
        return

    csv_files = [f for f in os.listdir(csv_folder) if f.endswith('.csv')]
    rows_per_file = max_rows // len(csv_files)
    print(f"Found {len(csv_files)} CSV files | "
          f"Budget per file: {rows_per_file:,}")

    for filename in csv_files:
        file_path = os.path.join(csv_folder, filename)
        try:
            df = pd.read_csv(file_path)
            if 'uid' in df.columns:
                df = df.drop(columns=['uid'])
            total_rows += len(df)

            if mode == "normal":
                filtered = df[df['type'] == 'normal']
                final_data.append(filtered)
                print(f"  {filename}: {len(filtered):,} normal rows")

            else:
                # Sample per file before appending — never load 22M into memory
                if len(df) > rows_per_file:
                    n_types   = df['type'].nunique()
                    per_type  = rows_per_file // n_types
                    sampled = []
                    for t, group in df.groupby('type'):
                        sampled.append(group.sample(n=min(len(group), per_type), random_state=42))
                    filtered = pd.concat(sampled, ignore_index=True)
                else:
                    filtered = df

                final_data.append(filtered)
                print(f"  {filename}: {len(filtered):,} / {len(df):,} rows sampled")

        except Exception as e:
            print(f"  ERROR {filename}: {e}")

    print(f"\nTotal raw rows processed: {total_rows:,}")

    if final_data:
        combined_df = pd.concat(final_data, ignore_index=True)

        if 'ts' in combined_df.columns:
            combined_df = combined_df.sort_values('ts').reset_index(drop=True)
            print("Sorted by timestamp")

        # print(f"\nFinal distribution:")
        # print("-----------------",combined_df.head())
        # print(combined_df['type'].value_counts())
        # print(f"Total rows: {len(combined_df):,}")
        combined_df.to_csv(output_file, index=False)

        print(f"Saved to {output_file}")
    else:
        print("No data found.")

combine_or_extract_normal_csv2(csv_folder, mode="all", max_rows=1_000_000)
```
